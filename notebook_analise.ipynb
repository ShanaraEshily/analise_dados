Estamos criando uma collection de dados para gerar um Dataframe

```python
dados = {
    'ID': [1, 2, 3, 4, 5],
    'Nome': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'CPF': ['123.456.789-00', '987.654.321-00', '111.222.333-44', '555.666.777-88', '999.888.777-66'],
}
```

Tranformando a collection em um DataFrame

```python
dt = pd.DataFrame(dados)
```

Imprime na tela o DataFrame

```python
dt.head()
```

Realizando transformação nos nomes das pessoas (UPPER)

```python
dt['Nome'] = dt['Nome'].str.upper()
```

```python
dt.head()
```

Filtrando os IDs 2, 3 e 4
<br>
(Data da OBS: 19/02/2026) Obs.: Sempre que for necessário realizar o filtro nos dados, armazenar o resultado em uma variável separada. Isso evita que os dados originais se percam.

```python
dt_filtrado = dt[(dt['ID'] >= 2) & (dt['ID'] <= 4)]
```
