```python
import pandas as pd
import mysql.connector
```

```python
connector = mysql.connector.connect(
    host = "localhost",
    port = 3307,
    user = "root",
    password = "senac",
    database = "ecommerce_etl"
)
```

Conecta no BD e realiza a extração dos dados utilizando uma Querry SQL

```python
querry = """
select c.nome, c.email, c.cidade,
	p.quantidade, p.data_pedido,
    pr.nome_produto, pr.preco
from clientes c
inner join pedidos p on c.id_cliente = p.id_cliente
inner join produtos pr on p.id_produto = pr.id_produto
"""
```

```python
dt = pd.read_sql(querry, connector)
dt.head()
```

Tranformação dos dados é o processo de converter dados de um formato ou estrutura para outro, a fim de torná-los mais adequados para análise, visualização ou armazenamento. Isso pode envolver várias operações, como limpeza de dados, normalização, agregação, filtragem e erniquecimento. A transformação de dados é uma etapa crucialem processos ETL (Extract - Tranform - Load)

```python
dt['email'] = dt['email'].fillna('em branco')
dt.head()
```

```python
dt['quantidade'] = dt['quantidade'].fillna(1)
dt.head()
```

Verificação do tipo do dataframe, para validar os tipos dos dados armazenados.<br>
Caso os tipos não estejam corretos, será necessário converter para o tipo correto.

```python
dt.info()
```

```python
dt['data_pedido'] = pd.to_datetime(dt['data_pedido'])
dt.info()
```

```python
dt['data_pedido'] = dt['data_pedido'].fillna(pd.Timestamp('2024-01-01'))
dt.head(20)
```

```python
dt['faturamento'] = dt['quantidade'] * dt['preco']
dt
```

```python
dt.to_excel('ecommerce_faturamento.xlsx', index = False)
```
